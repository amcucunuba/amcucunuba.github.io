```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42) 
table = pd.DataFrame()
table['Date'] = pd.date_range(start='2024-02-15', end='2025-06-30', freq='D')

table['ID_Customer'] = np.random.randint(10, 300, size=len(table))

products = [
    'Pistacho_cookie',
    'Redvelvet_cookie',
    'Chesscakelemon_cookie',
    'Biscoft_cookie',
    'Choconutella_cookie',
    'Chocomix_cookie',
    'Mini_cookies'
]

table['Product'] = np.random.choice(products, size=len(table))

product_prices = {
    'Pistacho_cookie': 5.5,
    'Redvelvet_cookie': 5.0,
    'Chesscakelemon_cookie': 5.5,
    'Biscoft_cookie': 4.5,
    'Choconutella_cookie': 4.0,
    'Chocomix_cookie': 4.0,
    'Mini_cookies': 2.0
}

product_costs = {
    'Pistacho_cookie': 2.5,
    'Redvelvet_cookie': 2.2,
    'Chesscakelemon_cookie': 2.6,
    'Biscoft_cookie': 2.0,
    'Choconutella_cookie': 1.8,
    'Chocomix_cookie': 1.9,
    'Mini_cookies': 3.8
}

table['price_per_unit'] = table['Product'].map(product_prices) 

table['cost_per_unit'] = table['Product'].map(product_costs)

table['profit_per_unit'] = table['price_per_unit'] - table['cost_per_unit']

table['Units_sold'] = np.random.randint(4, 12, size=len(table))

table['Total_sell'] = table['Units_sold'] * table['price_per_unit']

table['Total_cost'] = table['Units_sold'] * table['cost_per_unit']

table['Total_profit'] = table['Total_sell'] - table['Total_cost']

table
```

```python
table = table.sort_values(by="Date")

table['Customer_type'] = table.duplicated(subset='ID_Customer', keep='first')\
                                .map({False: 'New', True: 'Returning'})
```

```python
dates = pd.date_range(start='2024-02-15', end='2025-06-30', freq='M')

# Simulate Month operational costs
costs = pd.DataFrame({
    'Date': dates,
    'flour': np.random.uniform(10, 15, len(dates)),
    'butter': np.random.uniform(8, 12, len(dates)),
    'sugar': np.random.uniform(6, 10, len(dates)),
    'cream_cheese': np.random.uniform(0, 8, len(dates)),
    'lemon': np.random.uniform(1.5, 2.5, len(dates)),
    'coconut_scraper': np.random.uniform(0, 3, len(dates)),
    'cream_nutella': np.random.uniform(0, 10, len(dates)),
    'cream_pistacho': np.random.uniform(0, 15, len(dates)),
    'electricity': np.random.uniform(5, 9, len(dates)),
    'paper_trays': np.random.uniform(0, 3.5, len(dates)),
    'cookie_box': np.random.uniform(0, 30, len(dates)),
    'water': np.random.uniform(8, 15, len(dates)),
    'dish_soap': np.random.uniform(0, 2.5, len(dates)),
    'publicity': np.random.uniform(0, 15, len(dates)),
})

# Total cost per day
costs['total_month_cost'] = costs.drop(columns=['Date']).sum(axis=1)

# Preview
print(costs.head())
```
